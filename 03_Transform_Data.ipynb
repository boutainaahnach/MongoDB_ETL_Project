```python
import pandas as pd
import requests
import sqlite3
```

```python
# CSV
sales_df = pd.read_csv("../data/sales.csv")

# API
url = "https://fakestoreapi.com/products"
api_df = pd.DataFrame(requests.get(url).json())

# SQLite
conn = sqlite3.connect("../data/company.db")
employees_df = pd.read_sql_query("SELECT * FROM employees", conn)
```

```python
print("Avant :", len(sales_df))

# Supprimer les doublons
sales_df = sales_df.drop_duplicates()

# Supprimer les valeurs manquantes
sales_df = sales_df.dropna()

# Convertir la date
sales_df["date"] = pd.to_datetime(sales_df["date"])

print("Après :", len(sales_df))

sales_df.head()
```

```python
api_df = api_df[["id", "title", "price", "category"]]

api_df = api_df.rename(columns={
    "title": "product"
})

api_df = api_df.drop_duplicates()

api_df = api_df.dropna()

api_df.head()
```
