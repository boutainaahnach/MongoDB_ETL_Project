```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["ETL_Project"]

sales = db["sales"]
products = db["products"]
employees = db["employees"]
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$product",
            "Total_Quantity": {"$sum": "$quantity"}
        }
    },
    {
        "$sort": {"Total_Quantity": -1}
    }
]

for doc in sales.aggregate(pipeline):
    print(doc)
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$department",
            "Nombre": {"$sum": 1}
        }
    }
]

for doc in employees.aggregate(pipeline):
    print(doc)
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$category",
            "Prix_Moyen": {"$avg": "$price"}
        }
    }
]

for doc in products.aggregate(pipeline):
    print(doc)
```
