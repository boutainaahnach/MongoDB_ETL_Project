```python
from pymongo import MongoClient
import pandas as pd
import sqlite3
import requests
```

```python
# CSV
sales_df = pd.read_csv("../data/sales.csv")

# API
url = "https://fakestoreapi.com/products"
api_df = pd.DataFrame(requests.get(url).json())

# SQLite
conn = sqlite3.connect("../data/company.db")
employees_df = pd.read_sql_query("SELECT * FROM employees", conn)
```

```python
sales_df = sales_df.drop_duplicates().dropna()
sales_df["date"] = pd.to_datetime(sales_df["date"])

api_df = api_df[["id", "title", "price", "category"]]
api_df = api_df.rename(columns={"title": "product"})
api_df = api_df.drop_duplicates().dropna()

employees_df = employees_df.drop_duplicates().dropna()
```

```python
client = MongoClient("mongodb://localhost:27017/")

db = client["ETL_Project"]
```

```python
sales_collection = db["sales"]
products_collection = db["products"]
employees_collection = db["employees"]
```

```python
sales_collection.delete_many({})

sales_collection.insert_many(
    sales_df.to_dict("records")
)

print("Sales chargées.")
```

```python
products_collection.delete_many({})

products_collection.insert_many(
    api_df.to_dict("records")
)

print("Products chargés.")
```

```python
employees_collection.delete_many({})

employees_collection.insert_many(
    employees_df.to_dict("records")
)

print("Employees chargés.")
```

```python
print("Sales :", sales_collection.count_documents({}))
print("Products :", products_collection.count_documents({}))
print("Employees :", employees_collection.count_documents({}))
```
